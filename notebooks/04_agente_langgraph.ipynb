# 🛒 Shopper Assistant Agent — LangGraph

Versión del agente implementada con **LangGraph**.

## Arquitectura del grafo

```
[START] → [parsear_intencion] → [filtrar_catalogo] → [generar_canasta] → [END]
```

| Nodo | Responsabilidad |
|------|-----------------|
| `parsear_intencion` | GPT extrae presupuesto, dieta y restricciones nutricionales del texto libre |
| `filtrar_catalogo` | Filtra el DataFrame por precio, dieta y nutrición; muestrea productos diversos |
| `generar_canasta` | GPT selecciona productos y genera la canasta con justificación nutricional |
